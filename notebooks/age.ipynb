```python
import wfdb
from tqdm import tqdm

ages = []
for file in tqdm(cpsc['file_name']):
    path = '/home/datasets/CPSC2018/'
    full_path = path + file
    record = wfdb.rdheader(full_path)
    comment = record.__dict__['comments']
    #print(comment)
    for c in comment:
        if 'age' in c.lower():
            try:
                if int(c[4:].strip()) < 0:
                    continue
                ages.append(int(c[4:].strip()))
                break
            except ValueError:
                continue
```

```python
# get all the predictions
import torch
from tqdm import tqdm


def eval_model(model, dataloader):
    model.eval()
    model.to('cuda')
    outputs_list = []
    ages_list = []
    for batch in tqdm(dataloader):
        inputs = batch['signals'].to('cuda')
        ages = batch['age']
        with torch.no_grad():
            outputs = model(inputs)
            outputs_list.extend(outputs.flatten().cpu().numpy())
            ages_list.extend(ages.cpu().numpy())
        # process outputs
    return outputs_list, ages_list
    # process outputs
```

```python
import numpy as np
import seaborn as sns
from scipy import stats
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import os 
from matplotlib.gridspec import GridSpecFromSubplotSpec

def hex_plot(ax, ages, outputs, task, extension='svg'):
    sns.set_theme(style="ticks")
    
    # Calculate stats
    pearson_r, p_value = stats.pearsonr(outputs, ages)
    r2 = pearson_r ** 2

    g = sns.jointplot(x=outputs, y=ages, kind="hex", color="#4CB391", marginal_kws=dict(bins=25, fill=True), joint_kws=dict(gridsize=25, mincnt=1))
    g.set_axis_labels("Predicted Age", "Actual Age", fontsize=22)

    plt.xticks(fontsize=18, fontweight='medium')
    plt.yticks(fontsize=18, fontweight='medium')

    # Add stats legend to joint axis
    stats_label = f"r = {pearson_r:.3f}\n$R^2$ = {r2:.3f}"
    patch = mpatches.Patch(color='none', label=stats_label)
    g.ax_joint.legend(
        handles=[patch],
        title_fontsize=20,
        title=task,
        fontsize=18,
        loc='lower right',
        frameon=False,
        framealpha=0.8,
        edgecolor='gray'
    )
    os.makedirs("figs", exist_ok=True)  # creates the folder if it doesn't exist
    plt.savefig(f"figs/age_hex_plot_{task}_2.{extension}", dpi=300, bbox_inches='tight')

def joint_hex_subplot(fig, outer_spec, ages, outputs, title, palette):
    sns.set_theme(style="ticks")
    gs = GridSpecFromSubplotSpec(
        2, 2,
        subplot_spec=outer_spec,
        width_ratios=[4, 1],
        height_ratios=[1, 4],
        wspace=0.,
        hspace=0.
    )

    ax_top = fig.add_subplot(gs[0, 0])
    ax_right = fig.add_subplot(gs[1, 1])
    ax_joint = fig.add_subplot(gs[1, 0])

    sns.histplot(outputs, bins=25, ax=ax_top, color="#4CB391",  fill=True)
    sns.histplot(y=ages, bins=25, ax=ax_right, color="#4CB391", fill=True)
    ax_joint.hexbin(
        outputs,
        ages,
        gridsize=25,
        mincnt=1,
        cmap=palette
    )

    ax_joint.tick_params(axis='both', labelsize=18)

    ax_top.set_title(title, fontsize=22)

    ax_top.axis("off")
    ax_right.axis("off")

    return ax_joint
```
