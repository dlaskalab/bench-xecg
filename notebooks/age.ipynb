```python
cpsc = pd.read_csv('/media/Volume/data/CPSC2018/exams_filtered.csv')
cpsc.head()
```

```python
import wfdb
from tqdm import tqdm

ages = []
for file in tqdm(cpsc['file_name']):
    path = '/media/Volume/data/CPSC2018/'
    full_path = path + file
    record = wfdb.rdheader(full_path)
    comment = record.__dict__['comments']
    print(comment)
    for c in comment:
        if 'age' in c.lower():
            try:
                if int(c[4:].strip()) < 0:
                    continue
                ages.append(int(c[4:].strip()))
                break
            except ValueError:
                continue
```

```python
import numpy as np
# print min age cpsc
print(f"Min age CPSC: {np.min(ages)}")
# print max age cpsc
print(f"Max age CPSC: {np.max(ages)}")  
```

```python
# plot dist of ages cpsc
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111)
ax.hist(ages, bins=20, color='blue', alpha=0.7)
ax.set_title('Age Distribution of CPSC Patients')
ax.set_xlabel('Age')
ax.set_ylabel('Frequency')
plt.show()
```

```python
# loading the xlstm model
import utils.utils as utils
from models.classification import xLSTMClassification

# embedding_size: 1024
# encoder_type: vanilla
# xlstm_config: ["s","s","m","m","s", "s","m","m","s"]
# checkpoint: pretrain-xLSTM/0rgc8fun/checkpoints/epoch=49-step=504700.ckpt
# use_final_layer_norm: false
# cls_normalization: null

config = utils.parse_config('configs/train_age_run_config.yaml', 'config_defaults/train_age_defaults.yaml')
config.embedding_size = 1024
config.encoder_type = 'vanilla'
config.xlstm_config = ["s","s","m","m","s", "s","m","m","s"]
config.checkpoint = 'train-age/6ubu3eff/checkpoints/epoch=14-step=7905.ckpt'
config.use_final_layer_norm = False
config.cls_normalization = None
config.num_workers = 32

# load age default config
model = utils.get_base_model(config)
```

```python
from dataset.generic_utils import get_transforms, make_collate_fn_task
import dataset.ptb_xl as ptbxl

from torch.utils.data import DataLoader


test_ptbxl = ptbxl.ECGPTBXLAgeDataset(config, split='all', global_augmentations=get_transforms(config, split='test'))
test_ptbxl_dl = DataLoader(test_ptbxl, batch_size=config.batch_size, shuffle=False, num_workers=config.num_workers, collate_fn=make_collate_fn_task(config, key_label='age'))
```

```python
# get all the predictions
import torch
from tqdm import tqdm


def eval_model(model, dataloader):
    model.eval()
    model.to('cuda')
    outputs_list = []
    ages_list = []
    for batch in tqdm(dataloader):
        inputs = batch['signals'].to('cuda')
        ages = batch['age']
        with torch.no_grad():
            outputs = model(inputs)
            outputs_list.extend(outputs.flatten().cpu().numpy())
            ages_list.extend(ages.cpu().numpy())
        # process outputs
    return outputs_list, ages_list
    # process outputs
```

```python
outputs_list_ptb_xl, ages_list_ptb_xl = eval_model(model, test_ptbxl_dl)
```

```python
print(outputs_list_ptb_xl[:10])
print(ages_list_ptb_xl[:10])
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

def hex_plot(ages, outputs, task, extension='svg'):
    sns.set_theme(style="ticks")
    g = sns.jointplot(x=outputs, y=ages, kind="hex", color="#4CB391", marginal_kws=dict(bins=25, fill=True), joint_kws=dict(gridsize=25, mincnt=1))
    g.set_axis_labels("Predicted Age", "Actual Age", fontsize=22)

    # set xthicks font
    plt.xticks(fontsize=18, fontweight='medium')
    plt.yticks(fontsize=18, fontweight='medium')

    # legend
    plt.legend(title=task, title_fontsize=20, fontsize=14, loc='lower right', frameon=False)
    plt.savefig(f"figs/age_hex_plot_{task}.{extension}", dpi=300, bbox_inches='tight')

hex_plot(ages_list_ptb_xl, outputs_list_ptb_xl, 'PTB-XL')
```
